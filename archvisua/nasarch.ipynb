```python
import sys
sys.path.append('../')
from pycore.tikzeng import *

# Define architecture
arch = [
    to_head('..'),
    to_cor(),
    to_begin(),

    # First Conv Block
    to_Conv("conv1", 32, 8, offset="(0,0,0)", to="(0,0,0)", height=32, depth=32, width=2),
    to_Pool("pool1", offset="(0,0,0)", to="(conv1-east)", height=28, depth=28, width=1),

    to_Conv("conv2", 32, 32, offset="(1,0,0)", to="(pool1-east)", height=32, depth=32, width=3),
    to_Pool("pool2", offset="(0,0,0)", to="(conv2-east)", height=28, depth=28, width=1),

    to_Conv("conv3", 32, 16, offset="(1,0,0)", to="(pool2-east)", height=32, depth=32, width=2),
    to_Pool("pool3", offset="(0,0,0)", to="(conv3-east)", height=14, depth=14, width=1),

    to_Conv("conv4", 32, 64, offset="(1,0,0)", to="(pool3-east)", height=16, depth=16, width=3),
    to_Pool("pool4", offset="(0,0,0)", to="(conv4-east)", height=14, depth=14, width=1),

    to_Conv("conv5", 32, 64, offset="(1,0,0)", to="(pool4-east)", height=16, depth=16, width=3),
    to_Pool("pool5", offset="(0,0,0)", to="(conv5-east)", height=14, depth=14, width=1),

    to_Conv("conv6", 32, 16, offset="(1,0,0)", to="(pool5-east)", height=8, depth=8, width=2),
    to_Pool("pool6", offset="(0,0,0)", to="(conv6-east)", height=4, depth=4, width=1),

    to_Conv("conv7", 32, 16, offset="(1,0,0)", to="(pool6-east)", height=8, depth=8, width=2),
    to_Pool("pool7", offset="(0,0,0)", to="(conv7-east)", height=4, depth=4, width=1),

    to_Conv("conv8", 32, 22, offset="(1,0,0)", to="(pool7-east)", height=4, depth=4, width=2),
    to_Pool("pool8", offset="(0,0,0)", to="(conv8-east)", height=2, depth=2, width=1),

    # Fully Connected Layers (Using to_ConvSoftMax to represent Dense layers)
    to_ConvSoftMax("fc1", 198, offset="(1,0,0)", to="(pool8-east)", height=3, depth=3, width=5, caption="FC 1"),
    to_ConvSoftMax("fc2", 32, offset="(1,0,0)", to="(fc1-east)", height=2, depth=2, width=2, caption="FC 2"),
    to_ConvSoftMax("fc3", 32, offset="(1,0,0)", to="(fc2-east)", height=2, depth=2, width=2, caption="FC 3"),

    # Final Softmax Output
    to_SoftMax("softmax", 10, offset="(1,0,0)", to="(fc3-east)", caption="Softmax Output"),

    to_end()
]

def main():
    namefile = str(sys.argv[0]).split('.')[0]
    to_generate(arch, namefile + '.tex')

if __name__ == '__main__':
    main()
```
