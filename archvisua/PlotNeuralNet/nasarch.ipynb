```python
import sys
sys.path.append('../')
from pycore.tikzeng import *

def to_FullyConnected(name, s_filer=" ", n_filer=" ", offset="(0,0,0)", to="(0,0,0)", width=1.5, height=3, depth=25, opacity=0.8, caption=" ", zlabelposition='midway'):
    return r"""
\pic[shift={"""+ offset +"""}] at """+ to +""" 
    {Box={
        name=""" + name +""",
        caption=""" +caption + """,
        xlabel={{ """+ '"'+str(n_filer) +'", "dummy"'+ """ }},
        zlabel="""+ str(s_filer) +""",
        zlabelposition="""+zlabelposition+""",
        fill=\FcColor,
        bandfill=\FcReluColor,
        opacity="""+ str(opacity) +""",
        height="""+ str(height) +""",
        width="""+ str(width) +""",
        depth="""+ str(depth) +"""
        }
    };
"""

# Define architecture
arch = [
    to_head('..'),
    to_cor(),
    to_begin(),

    # Input Layer
    to_input("cat.jpg", width=32, height=32, name="input"),

    # Convolutional and Pooling Layers Based on Computed Shapes
    to_Conv("conv1", 30, 8, offset="(0,0,0)", to="(0,0,0)", height=30, depth=30, width=2, caption="conv1"),
    to_Pool("pool1", offset="(0,0,0)", to="(conv1-east)", height=29, depth=29, width=1),

    to_Conv("conv2", 29, 32, offset="(1,0,0)", to="(pool1-east)", height=29, depth=29, width=3, caption="conv2"),
    to_Pool("pool2", offset="(0,0,0)", to="(conv2-east)", height=28, depth=28, width=1),

    to_Conv("conv3", 28, 16, offset="(1,0,0)", to="(pool2-east)", height=28, depth=28, width=2, caption="conv3"),
    to_Pool("pool3", offset="(0,0,0)", to="(conv3-east)", height=14, depth=14, width=1),

    to_Conv("conv4", 14, 64, offset="(1,0,0)", to="(pool3-east)", height=14, depth=14, width=3, caption="conv4"),
    to_Pool("pool4", offset="(0,0,0)", to="(conv4-east)", height=7, depth=7, width=1),

    to_Conv("conv5", 7, 64, offset="(1,0,0)", to="(pool4-east)", height=7, depth=7, width=3, caption="conv5"),
    to_Pool("pool5", offset="(0,0,0)", to="(conv5-east)", height=3, depth=3, width=1),

    to_Conv("conv6", 3, 16, offset="(1,0,0)", to="(pool5-east)", height=3, depth=3, width=2, caption="conv6"),
    to_Pool("pool6", offset="(0,0,0)", to="(conv6-east)", height=1, depth=1, width=1),

    # Fully Connected Layers
    to_FullyConnected("fc1", 198, offset="(1.25,0,0)", to="(pool6-east)", width=1, height=1, depth=32, caption="fc1"),
    to_FullyConnected("fc2", 32, offset="(1.25,0,0)", to="(fc1-east)", width=1, height=1, depth=16, caption="fc2"),
    to_FullyConnected("fc3", 32, offset="(1.25,0,0)", to="(fc2-east)", width=1, height=1, depth=16, caption="fc3"),
    
    # Final Softmax Output
    to_SoftMax("softmax", 10, offset="(1,0,0)", to="(fc3-east)", caption="Softmax"),
    
    # Connections
    to_connection("pool1", "conv2"),
    to_connection("pool2", "conv3"),
    to_connection("pool3", "conv4"),
    to_connection("pool4", "conv5"),
    to_connection("pool5", "conv6"),
    to_connection("pool6", "fc1"),
    to_connection("fc1", "fc2"),
    to_connection("fc2", "fc3"),
    to_connection("fc3", "softmax"),
    
    to_end()
]

def main():
    namefile = str(sys.argv[0]).split('.')[0]
    to_generate(arch, namefile + '.tex')

if __name__ == '__main__':
    main()
```

```python
import torch
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

# Load CIFAR-10 dataset
transform = transforms.ToTensor()
dataset = torchvision.datasets.CIFAR10(root="./", train=True, download=True, transform=transform)

# CIFAR-10 class labels
classes = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']

# Find the first cat image
for i in range(len(dataset)):
    image, label = dataset[i]
    if classes[label] == 'cat':
        # Convert Tensor to NumPy
        np_image = image.permute(1, 2, 0).numpy() * 255
        np_image = np_image.astype(np.uint8)
        
        # Save as an image
        img = Image.fromarray(np_image)
        img.save("cat_image.png")
        print("Cat image saved as 'cat_image.png'")
        break
```
